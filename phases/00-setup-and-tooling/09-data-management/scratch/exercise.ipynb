# Stream the c4 dataset and count how many examples you can process in 10 seconds

```python
import time

dataset = load_dataset(
    'allenai/c4',
    'en',
    streaming=True,
    split='train'
)

start_time = time.time()
count = 0

for count, example in enumerate(dataset):
    if time.time() > start_time + 10:
        break

print(f"Example processed: {count}")
```

# Convert a dataset to Parquet and compare the file size to CSV

```python
import os

dataset = load_dataset("nyu-mll/glue", "mrpc", split='train')

dataset.to_csv('glue_train.csv')
dataset.to_parquet("glue_train.parquet")

csv_size = os.path.getsize('glue_train.csv') / (1024 * 1024)
parquet_size = os.path.getsize('glue_train.parquet') / (1024 * 1024)

print(f"CSV Size: {csv_size:.2f} MB")
print(f"Parquet Size: {parquet_size:.2f} MB")
```

# Create a 70/15/15 train/val/test split with a fixed seed and verify the sizes

```python
dataset = load_dataset("nyu-mll/glue", "mrpc", split='train')

split = dataset.train_test_split(test_size=0.3, seed=42)
train_val = split['test'].train_test_split(test_size=0.5, seed=42)

train_dataset = split["train"]
val_dataset = train_val["train"]
test_dataset = train_val["test"]

print(f"Train      : {len(train_dataset)}")
print(f"Validation : {len(val_dataset)}")
print(f"Test       : {len(test_dataset)}")
```
